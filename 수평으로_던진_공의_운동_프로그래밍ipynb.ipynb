```python
import math

PI = 3.141592
g = -9.8

v0 = float(input("초기 속도를 입력하세요(m/s) : "))

theta_degree = float(input("각도를 입력하세요(°) : "))

# 0. 각도를 라디안으로 변환
theta_radian = math.radians(theta_degree)

# 1. x,y방향의 초기속도를 구하여라
v0x = v0 * math.cos(theta_radian)
print("x의 초기 속도 :", v0x, "m/s")

v0y = v0 * math.sin(theta_radian)
print("y의 초기 속도 :", v0y, "m/s")


# 2. 최고점의 높이를 구하여라
H = (v0y*v0y) / (2 * -g)
print("최고점의 높이 :", H, "m")

# 3. 최고점에 도달하는 시간을 구하여라
T = v0y / -g
print("최고점 도달 시간 :", T, "s")

# 4. 수평으로 이동한 최고 거리를 구하여라
R = v0x * 2 * T
print("수평 최고 거리 :", R, "m")

# 5. 지면에 닿을때의 Vx, Vy를 구하고 V의 벡터 성분을 구하여라
vx = v0x
print("지면에 닿을 때의 vx :", vx, "m/s")

vy = -v0y
print("지면에 닿을 때의 vy :", vy, "m/s")

V = math.sqrt(vx**2 + vy**2)
print("지면에 닿을때 V 백터 크기 :", V, "m/s")
```
