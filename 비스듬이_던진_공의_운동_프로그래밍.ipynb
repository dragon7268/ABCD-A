```python
import math

# 초기값들
v0 = 100  # 초기 속도 (m/s)
theta_deg = 37  # 발사 각도 (도)
theta_rad = math.radians(theta_deg)  # 라디안으로 변환
g = -9.8  # 중력 가속도 (음수로 설정)
y0 = 140  # 초기 높이 (m)

# 속도 성분
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)

# (a) 수면에 도달하는 시간 (2차 방정식의 해 구하기)
# y = y0 + v0y * t + 0.5 * g * t^2 = 0
# 즉: 0.5 * g * t^2 + v0y * t + y0 = 0
a = 0.5 * g
b = v0y
c = y0

# 근의 공식 사용
discriminant = b**2 - 4 * a * c
if discriminant < 0:
    raise ValueError("실근이 존재하지 않습니다.")

t1 = (-b + math.sqrt(discriminant)) / (2 * a)
t2 = (-b - math.sqrt(discriminant)) / (2 * a)

# 양의 시간 선택
t = max(t1, t2)

# (b) 수평 거리
x = v0x * t

# (c) 수면 도달 순간 속도 성분
vx = v0x
vy = v0y + g * t  # g는 음수이므로 덧셈

# (d) 속도의 각도
angle_rad = math.atan2(vy, vx)
angle_deg = math.degrees(angle_rad)

# 결과 출력
print(f"(a) 수면에 도달하는 시간: {t:.2f} s")
print(f"(b) 수평 거리: {x:.2f} m")
print(f"(c) 수면 도달 시 속도 성분: vx = {vx:.2f} m/s, vy = {vy:.2f} m/s")
print(f"(d) 수면 도달 시 속도의 각도: {angle_deg:.2f}°")
```
